Task 1. Color Space Conversion  

Using your low-light photo, convert it to Grayscale, HSV, and RGB (remember OpenCV loads as BGR by default). Display all versions side by side. 

Then, looking specifically at your own photo, note which HSV channel best preserves detail in the darkest part of the frame, and why you think that's the case for this particular shot. 

```python
import cv2
from IPython.display import Image
import matplotlib.pyplot as plt
```

```python
img = cv2.imread("raw_captures\Sen_Dark.jpg")
```

```python
img_gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
img_hsv = cv2.cvtColor(img,cv2.COLOR_BGR2HSV)
img_rgb = cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
```

```python
H, S, V = cv2.split(img_hsv)
```

```python
plt.Figure(figsize=(18,8))

plt.subplot(2,4,1)
plt.imshow(img_gray)
plt.title("Grayscale")
plt.axis("Off")

plt.subplot(2,4,2)
plt.imshow(img_rgb)
plt.title("RGB")
plt.axis("Off")

plt.subplot(2,4,3)
plt.imshow(H,cmap="gray")
plt.title("Hue")
plt.axis("Off")

plt.subplot(2,4,4)
plt.imshow(S, cmap="gray")
plt.title("Saturation")
plt.axis("Off")


plt.subplot(2,4,5)
plt.imshow(V,cmap="gray")
plt.title("Value")
plt.axis("Off")


plt.tight_layout()
plt.show()
```

In my low-light image, the Value (V) channel preserved the most detail in the darkest parts of the frame. The V channel represents pixel brightness, so even though the scene is dark, small differences in illumination are still visible. The Hue channel lost detail because color information becomes unreliable under very low light, and the Saturation channel showed limited information since the colors in shadowed regions were not strongly saturated.
