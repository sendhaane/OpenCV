Task 9. Real-Time Video Processing   

Using your short video clip, apply a real-time effect of your choice frame by frame (grayscale, edge detection, face detection, or a color filter) and save the output with cv2.VideoWriter. Screen-record yourself running it, not just the final clean output, and note one bug you hit along the way and how you fixed it. 

```python
import cv2
```

```python
cap = cv2.VideoCapture("raw_captures/race_car.mp4")
```

```python
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS)
```

```python
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
```

```python
out = cv2.VideoWriter(
    "output_video.mp4",
    fourcc,
    fps,
    (width, height),
    False     

)
```

```python
while True:

    ret, frame = cap.read()

    if not ret:
        break

    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    
    edges = cv2.Canny(gray, 50, 150)

    
    cv2.imshow("Edge Detection", edges)

    
    out.write(edges)


    if cv2.waitKey(1) == ord('q'):
        break

cap.release()
out.release()

cv2.destroyAllWindows()
```

I applied Canny Edge Detection to each frame of the video. Every frame was converted to grayscale before applying edge detection, and the processed frames were displayed in real time and saved to a new video using cv2.VideoWriter.

Bug: Initially, the output video was not saved correctly because the VideoWriter was created with the wrong color setting. The edge-detected frames are grayscale images, but the writer was expecting color frames.

Fix: I set the last parameter of cv2.VideoWriter to False, indicating that the output frames are grayscale. After this change, the video was saved correctly.
